# Flask: Diferencia entre Controller, Utils y Model

Este cuaderno tiene como objetivo explicar de manera clara la diferencia entre tres componentes clave en una arquitectura de aplicación Flask bien estructurada: **controllers**, **utils** y **models**.

---

## 1. ¿Qué es un Controller en Flask?

Un **controller** es una parte de la aplicación encargada de recibir las peticiones del navegador (rutas), procesar la lógica necesaria y retornar una respuesta adecuada (HTML, JSON, redirecciones, etc.).

### Características:
- Gestiona rutas (`@app.route` o `Blueprints`).
- Llama a servicios o modelos para procesar datos.
- Devuelve vistas o respuestas al usuario.

### Ejemplo:
```python
# controllers/usuario_controller.py
from flask import Blueprint, render_template
from models.usuario import Usuario

usuario_bp = Blueprint('usuario', __name__)

@usuario_bp.route("/usuarios")
def listar_usuarios():
    usuarios = Usuario.query.all()
    return render_template("usuarios.html", usuarios=usuarios)
```

---

## 2. ¿Qué es un módulo Utils?

Un archivo `utils.py` contiene **funciones auxiliares** o herramientas que pueden ser reutilizadas en múltiples partes de la aplicación. No tienen lógica de negocio ni gestionan rutas.

### Características:
- Funciones independientes.
- No se relacionan directamente con peticiones o respuestas.
- Mejoran la reutilización y limpieza del código.

### Ejemplo:
```python
# utils/formato.py
def formatear_fecha(fecha):
    return fecha.strftime('%d/%m/%Y')

def capitalizar_nombre(nombre):
    return nombre.strip().title()
```

---

## 3. ¿Qué es un Model?

En Flask (usualmente con SQLAlchemy), un **model** representa una tabla en la base de datos. Define los atributos de la entidad y permite hacer operaciones CRUD (crear, leer, actualizar, eliminar).

### Características:
- Se define como una clase que hereda de `db.Model`.
- Representa entidades de base de datos (como Usuario, Producto).
- Usado por controladores y servicios.

### Ejemplo:
```python
# models/usuario.py
from flask_sqlalchemy import SQLAlchemy

db = SQLAlchemy()

class Usuario(db.Model):
    __tablename__ = 'usuarios'

    id = db.Column(db.Integer, primary_key=True)
    nombre = db.Column(db.String(100))
    correo = db.Column(db.String(120), unique=True)
```

---

## 4. Comparación General

| Componente | Rol principal | Relación con Flask | Uso |
|------------|---------------|--------------------|-----|
| Controller | Manejar rutas y lógica de control | Sí, mediante `@route` o `Blueprint` | Interfaces con el navegador |
| Utils      | Funciones auxiliares reutilizables | No directamente | Limpieza de texto, formato de fechas |
| Model      | Representar datos y lógica de DB | Sí, con SQLAlchemy u ORM | Consultas y manipulación de datos |

---

## 5. Conclusión

- Usa **controllers** para organizar tu lógica web.
- Usa **utils** para separar funciones auxiliares reutilizables.
- Usa **models** para representar y consultar entidades de base de datos.

Una buena separación de responsabilidades ayuda a mantener tu aplicación organizada, escalable y fácil de mantener.
